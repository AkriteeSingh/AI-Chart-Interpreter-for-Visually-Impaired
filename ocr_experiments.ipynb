```python
from ultralytics import YOLO

def detect_chart_elements(image_path, model_path='best.pt', conf_threshold=0.4, iou_threshold=0.4):
    model = YOLO(model_path)
    results = model(image_path, conf=conf_threshold, iou=iou_threshold)
    detections = []
    for box in results[0].boxes:
        detections.append({
            "class": model.names[int(box.cls)],
            "bbox": [round(x, 2) for x in box.xyxy[0].tolist()],
            "confidence": round(float(box.conf), 4)
        })
    classes_found = [d["class"] for d in detections]
    chart_type = "bar" if "bar" in classes_found else "pie" if "pie" in classes_found else "unknown"
    return {"chart_type": chart_type, "detections": detections}
```

```python
def validate_pie_total(series):
    """Check if percentages sum close to 100. Does NOT touch individual slice confidence —
    just reports whether the chart-level total looks right, as separate metadata."""
    values = [s["value"] for s in series if s["value"] is not None]
    if not values:
        return False, None
    total = sum(values)
    total_ok = abs(total - 100) <= 15
    return total_ok, round(total, 1)
```

```python
def parse_percentage(text):
    if text is None:
        return None
    try:
        cleaned = text.replace("%", "").strip()
        value = float(cleaned)
        if value < 0 or value > 100:
            return None  # nonsensical for a percentage — OCR garbled this
        return value
    except ValueError:
        return None
```

```python
def extract_title(detections, image_path, reader):
    title_boxes = [d for d in detections if d["class"] == CLASS_TITLE]
    if not title_boxes:
        return None, 0.0
    best = max(title_boxes, key=lambda d: d["confidence"])
    cropped = crop_box(image_path, best["bbox"])
    text, ocr_conf = ocr_crop(reader, cropped)
    conf = combined_confidence(best["confidence"], ocr_conf if text else 1.0)
    return text, conf
```

```python
def process_pie_chart(image_path, detections, reader):
    """Pie charts: one 'pie' box for the whole chart; each 'label' box = one slice's on-slice text"""
    title_text, title_conf = extract_title(detections, image_path, reader)
    labels = [d for d in detections if d["class"] == CLASS_LABEL]

    series = []
    for label_box in labels:
        cropped = crop_box(image_path, label_box["bbox"])
        text, ocr_conf = ocr_crop(reader, cropped)

        conf = combined_confidence(label_box["confidence"], ocr_conf if text else 1.0)
        series.append({
            "category": text,   # will refine once we see real OCR output (e.g. "Product A: 25%")
            "value": parse_percentage(text),       
            "confidence": round(conf, 3),
            "low_confidence": bool(conf < 0.7 or text is None)
        })
    total_ok, total_value = validate_pie_total(series)  
    return {
        "chart_type": "pie",
        "title": title_text,
        "series": series,
        "metadata": {
            "num_series": len(series),
            "overall_confidence": round(sum(s["confidence"] for s in series) / len(series), 3) if series else 0,
            "percentages_sum_to_100": total_ok,
            "percentage_total": total_value
        }
    }
```

```python
def match_categories_to_bars(bars, categories):
    available = list(range(len(categories)))
    bar_order = sorted(range(len(bars)), key=lambda i: (bars[i]["bbox"][0] + bars[i]["bbox"][2]) / 2)

    matches = {}
    for bar_idx in bar_order:
        bar_cx = (bars[bar_idx]["bbox"][0] + bars[bar_idx]["bbox"][2]) / 2
        if not available:
            matches[bar_idx] = None
            continue
        best = min(available, key=lambda j: abs((categories[j]["bbox"][0] + categories[j]["bbox"][2]) / 2 - bar_cx))
        matches[bar_idx] = best
        available.remove(best)
    return matches
```

```python
from PIL import Image

# ---- Step 0: class names, confirmed directly from her model ----
CLASS_BAR = "bar"
CLASS_PIE = "pie"
CLASS_TITLE = "title"
CLASS_X_LABEL = "x_label"
CLASS_Y_LABEL = "y_label"
CLASS_LABEL = "label"  # confirmed singular


# ---- Step 1: crop a box out of the image ----
def crop_box(image_path, bbox):
    """bbox format: [x1, y1, x2, y2]"""
    img = Image.open(image_path)
    return img.crop((bbox[0], bbox[1], bbox[2], bbox[3]))


# ---- Step 2: run OCR on one cropped region ----
def ocr_crop(reader, cropped_img):
    """Returns (text, confidence) — picks the single best-confidence read in that crop"""
    import numpy as np
    result = reader.readtext(np.array(cropped_img))
    if not result:
        return None, 0.0  # nothing readable in this box
    # if OCR found multiple text pieces in one crop, join them, average confidence
    texts = [r[1] for r in result]
    confs = [r[2] for r in result]
    combined_text = " ".join(texts)
    avg_conf = sum(confs) / len(confs)
    return combined_text, avg_conf


# ---- Step 3: tell category labels apart from tick values ----
def classify_label_boxes(detections, image_height):
    """
    model calls both x-axis categories AND y-axis tick values 'label'.
    We guess which is which using position:
    - near the bottom of the image -> likely an x-axis category (e.g. "A")
    - near the left side -> likely a y-axis tick value (e.g. "50")
    This is a simple heuristic — refine later if it misclassifies on real charts.
    """
    categories = []
    tick_values = []
    for d in detections:
        if d["class"] != CLASS_LABEL:
            continue
        x1, y1, x2, y2 = d["bbox"]
        box_center_y = (y1 + y2) / 2
        box_center_x = (x1 + x2) / 2
        # crude rule: bottom third of image -> x-axis category
        if box_center_y > image_height * 0.7:
            categories.append(d)
        else:
            tick_values.append(d)
    return categories, tick_values


# ---- Step 4: calibrate pixel position -> real value, using tick values ----
def build_calibration(tick_detections, ocr_results_for_ticks):
    points = []
    for det, (text, conf) in zip(tick_detections, ocr_results_for_ticks):
        if text is None:  # OCR found nothing readable in this crop — skip it
            continue
        try:
            value = float(text.replace(",", "").strip())
            y1, y2 = det["bbox"][1], det["bbox"][3]
            pixel_y = (y1 + y2) / 2
            points.append((pixel_y, value))
        except ValueError:
            continue

    if len(points) < 2:
        return None

    points.sort()
    (y_low, val_low), (y_high, val_high) = points[0], points[-1]

    def pixel_to_value(pixel_y):
        if y_high == y_low:
            return None
        ratio = (pixel_y - y_low) / (y_high - y_low)
        return val_low + ratio * (val_high - val_low)

    return pixel_to_value


# ---- Step 5: combine confidence scores ----
def combined_confidence(detection_conf, ocr_conf=1.0):
    """Weakest-link approach — take the lowest of the two"""
    return float(min(detection_conf, ocr_conf))


# ---- Step 6: full pipeline ----
def process_chart(image_path, reader, model_path='best.pt'):
    from ultralytics import YOLO
    model = YOLO(model_path)
    img = Image.open(image_path)
    image_height = img.height

    results = model(image_path, conf=0.4, iou=0.4)
    detections = []
    for box in results[0].boxes:
        detections.append({
            "class": model.names[int(box.cls)],
            "bbox": [round(x, 2) for x in box.xyxy[0].tolist()],
            "confidence": round(float(box.conf), 4)
        })

    classes_found = [d["class"] for d in detections]
    chart_type = "bar" if CLASS_BAR in classes_found else "pie" if CLASS_PIE in classes_found else "unknown"

    if chart_type == "pie":
        return process_pie_chart(image_path, detections, reader)
        
    title_text, title_conf = extract_title(detections, image_path, reader)

    bars = [d for d in detections if d["class"] == CLASS_BAR]
    categories, tick_values = classify_label_boxes(detections, image_height)

    # OCR the tick values and categories
    tick_ocr = [ocr_crop(reader, crop_box(image_path, d["bbox"])) for d in tick_values]
    category_ocr = [ocr_crop(reader, crop_box(image_path, d["bbox"])) for d in categories]

    # build calibration from ticks
    calibrate = build_calibration(tick_values, tick_ocr)

    category_matches = match_categories_to_bars(bars, categories)
    # build final series: one entry per bar
    series = []
    for i, bar in enumerate(bars):
        bar_top_y = bar["bbox"][1]  # top edge of bar
        value = calibrate(bar_top_y) if calibrate else None
        category_text, category_conf = None, 0.0
        matched_idx = category_matches.get(i)
        if matched_idx is not None:
            category_text, category_conf = category_ocr[matched_idx]

        value_known = value is not None
        conf = combined_confidence(bar["confidence"], category_conf if category_text else 1.0)

        series.append({
            "category": category_text if category_text else None,
            "value": round(value, 2) if value_known else None,
            "confidence": round(conf, 3),
            "low_confidence": bool(conf < 0.7 or not value_known)
        })

    return {
        "chart_type": chart_type,
        "title": title_text,  
        "series": series,
        "metadata": {
            "num_series": len(series),
            "overall_confidence": round(sum(s["confidence"] for s in series) / len(series), 3) if series else 0
        }
    }
```
